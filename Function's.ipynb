#### Create a function for checking whether a number is prime or not 
           - function take one argument as number
           - return weather it is prime or not

```python
def prime_number(a):
    if a % 2 == 0:
        return False
    else:
        return True
```

```python
prime_number(10)
```

```python
prime_number(7)
```

```python
prime_number(17)
```

```python
def prime(num):
    # handle 1 and 0 using if
    for i in range(2,num):
        if num % 2 == 0:
            return False
            
    else:
        return True
```

```python
prime(17)
```

```python
prime(10)
```

```python
prime(1)
```
