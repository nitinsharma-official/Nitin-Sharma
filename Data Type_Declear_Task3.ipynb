## Question 9.

difference between ordered and unordered data types :-

    - ordered data type:-
        - Ordered data types store elements in a specific sequence, and this order is maintained.
        - Elements accessed by index or position 
        - List, Tuple, String
    - unordered data type:-
        - Unordered data types do not maintain any specific order of elements.
        - Elements accessed by key or value only
        - Set, Dictionary
