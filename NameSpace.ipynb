```python
x = 80
def fun1():
    global x
    x += 20
    print(x)
    def fun2():
        x = 20
        print(x)
    fun2()
fun1()
```

```python
x = 80
def fun1():          # enclosing / non local
    x = 20
    print(x)
    def fun2():      # local space
        nonlocal x
        x += 20
        print(x)
    fun2()
fun1()
```

```python
def extend_ls(value , ls = []):
    ls.append(value)
    return ls
```

```python
a = extend_ls(5)
b = extend_ls(10)
c = extend_ls(22)
print(a)
print(b)
print(c)
```

```python
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```
