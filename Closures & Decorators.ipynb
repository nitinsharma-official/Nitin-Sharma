## Closure
     - something that need to be closed
     - store the variable value of a function when the function clean the memory

```python
def outer():
    print("Hello user")
    def inner():
        x = 50
        print(x)
    return inner
```

```python
outer()
```

```python
v = outer()
```

```python
v()
```

## Decoraors
    - implementation of closure
    - to decorate a function without modifying the existing function

```python
def fun():
    return "Hello World"
```

```python
def decorator(fun):
    def wrapper():
        print("*"*40)
        print(fun())
        print("*"*40)
    return wrapper
```

```python
decorator(fun)()
```

```python
def greet():
    return "Good Morning"
```

```python
decorator(greet)()
```

```python
def name():
    return "My Name is Nitin Sharma"
```

```python
decorator(name)()
```

```python
def decorate(fun2):
    def fun1(a , b):
        print("!"*10)
        print(fun2(a ,b))
        print("^"*10)
    return fun1
```

```python
def add(a,b):
    return a + b
```

```python
decorate(add)(5,4)
```

```python
@decorate
def say_hi():
    return "hii"
```

```python
say_hi()
```
