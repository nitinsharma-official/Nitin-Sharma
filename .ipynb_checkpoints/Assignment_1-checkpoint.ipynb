## Question 3.

```python
english = {121, 1, 2, 3, 4, 5, 6, 7, 8, 9, 98, 78, 56, 12, 23, 43, 100}
hindi   = {123, 10, 1, 2, 3, 11, 21, 55, 6, 8, 34, 12, 45, 76, 89, 100}
french  = {1, 2, 23, 3, 4, 5, 6, 7, 8, 9, 56, 76, 87, 98, 23, 12, 44, 100}
chinese = {10, 1, 22, 2, 3, 11, 21, 55, 6, 8, 34, 12, 34, 56, 87, 671, 100}
at_least_one = english.union(hindi, french, chinese)
print("1. Total number of persons who have known at least one language:", len(at_least_one))
from collections import defaultdict

freq = defaultdict(int)

for person in at_least_one:
    if person in english:
        freq[person] += 1
    if person in hindi:
        freq[person] += 1
    if person in french:
        freq[person] += 1
    if person in chinese:
        freq[person] += 1
any_two = sum(1 for count in freq.values() if count == 2)
print("2. Total number of persons who have known any two languages:", any_two)
all_languages = english.intersection(hindi, french, chinese)
print("3. Total number of persons who have known all languages:", len(all_languages))
only_english = english - hindi - french - chinese
print("4. Total number of persons who have known only English language:", len(only_english))
```

## Question 5.

```python
player1={'name':'Virat Kohli', 'role':'bat', 'runs':112, '2':8,'4':10, '6':5, 'balls':109, 'field':0}
player2={'name':'du Plessis', 'role':'bat', 'runs':120,'2':3, '4':3, '6':8,'balls':112, 'field':0}
player3={'name':'Bhuvneshwar Kumar', 'role':'bowl', 'wkts':1, 'overs':10,'runs':71, 'field':1}
player4={'name':'Yuzvendra Chahal', 'role':'bowl', 'wkts':2, 'overs':10, 'runs':45, 'field':0}
player5={'name':'Kuldeep Yadav', 'role':'bowl', 'wkts':3, 'overs':10, 'runs':34,'field':0}

players=[player1 , player2 , player3 , player4 , player5 ] 
p={}

for i in players:
    score=0
    if i['role']=='bat':
        score += i['runs']//2                   
        if i['runs']>=50: score+=5
        if i['runs']>=100: score+=10            
        score += i['4']*1 + i['6']*2            
        strike_rate = (i['runs']/i['balls'])*100
        if strike_rate>=100: score+=10          
        score += i['field']*5                   
    
    elif i['role']=='bowl':
        score += i['wkts']*10
        if i['wkts']>=2: score+=5              
        eco = i['runs']/i['overs']
        if 3.5 <= eco <= 4.5: score+=4
        if 3 <= eco <= 5.5: score+=7
        score += i['field']*5
    
    p[i['name']] = score

for name,pt in p.items():
    print(name, ":", pt, "points")

lowest_player = min(p, key=p.get)
highest_player = max(p, key=p.get)

print(f"Lowest Points :- ", lowest_player, "-", p[lowest_player], "points")
print(f"Highest Points :- ", highest_player, "-", p[highest_player], "points")
```
