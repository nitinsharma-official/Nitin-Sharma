    - 18. Write a pyhton program to input electricity unit charges and calculate total electricity bill according to the given condition:
           - For first 50 units Rs. 0.50/ unit
           - For next 100 units Rs. 0.75/unit
           - For next 100 units Rs. 1.20/unit
           - For unit above 250 Rs.1.50/unit
           - An additional surcharge of 20% is added to the bill

```python
unit = float(input("Enter No. of unit"))
bill = 0
if unit <= 50:
    bill = unit * 0.50
elif unit <= 150:
    bill = (unit - 50) * 0.75 + 25
elif unit <= 250:
    bill = (unit - 150) * 1.20 + 100
elif unit <= 250:
    bill = (unit - 250) * 1.50 + 220
surcharge = bill * 0.20
Total_bill = bill + surcharge
print(f"your final bill is {Total_bill}")
```

## Linear Search

```python
ls = [3,4,5,6,7,44,55,66,33,12,23,5,6,8]
```

```python
search_element = int(input("Enter Your element to search"))
i = 0
while i < len(ls):
    if search_element == ls[i]:
        print(f"{search_element} found at position No. {i}")
        break
    i += 1
else:
    print(f"{search_element} is not in list")
```

```python
ls = [3,4,5,6,7,44,55,66,33,12,23,5,7,7,7,6,8]
```

```python
search_element = int(input("Enter Your element to search"))
i = 0
element = False
while i < len(ls):
    if search_element == ls[i]:
        print(f"{search_element} found at position No. {i}")
        element = True
    i += 1
if not element:
    print(f"{search_element} is not in list")
```

```python
search_element = int(input("Enter the element to search: "))
count = 0
for index in range(len(ls)):
    if ls[index] == search_element:
        print(f"{search_element} found at position No. {index}")
        count += 1

if count == 0:
    print(f"{search_element} is not in the list")
```

## Prime Number

```python
prime = int(input("Enter a Number"))
i = 2
while i < prime:
    if prime % i == 0:
        print("Not a prime number")
        break
    i += 1
else:
    print(" prime number")
```

```python
num = int(input("Enter a number"))
temp = num
rem = 0
rev = 0
while num:
    rem = num % 10
    rev = rev * 10 + rem
    num //= 10

if temp == rev:
    print(f"palindrome")
else:
    print("Not palindrome")
```

```python
password = random.choice(string.ascii_uppercase) + random.choice(string.punctuation) + random.choice(string.digits)
user_choice = int(input("Enter length of password you want ?"))
i = 1
while i <= user_choice - 3:
    password += random.choice(string.ascii_letters + string.digits + string.punctuation)
    i += 1
else:
    print(f"Length of password is {len(password)}")
    print(f"Your Random Password is {password}")
```
