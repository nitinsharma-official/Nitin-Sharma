## TypeCasting (Type conversion)
    - Typecasting is used in python to change the one data type into another data type.
    - Types of Typecasting
      - Implicit >> Automatically
      - Explicit >> Forcefully

## Implicit Typecasting (Automatically)

```python
a = 220    # int value
b = 110.25  # float
c = a + b    # float
print(c)
```

```python
type(a)
```

```python
type(b)
```

```python
type(c)
```

```python
a = 5  # int
b = 3j   # complex
c = a + b  # complex number
print(c)
```

```python
type(c)
```

```python
type(b)
```

```python
ls = [("Name", "Nitin"), ("Designation", "Data Analyst"), ("Salary", 75000)]

for i, (key, value) in enumerate(ls):
    if key == "Designation":
        # Replace "Data" with "Data Science" in the value
        new_value = value.replace("Data Analyst", "Data Science")
        ls[i] = (key, new_value)  # Update the tuple in the list
        break

print(ls)
```
