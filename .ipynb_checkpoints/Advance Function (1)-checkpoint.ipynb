## Map
    - map is a function which takes one function and apply that function on all over the iterable
    - syntax
       - map(function , iterable)

```python
ls = ["Simran","Meenakshi","Kirti","Ganeshu","Nitin","Abhishek"]
```

```python
for i in ls:
    print(i.upper())
```

```python
list(map(lambda a : a.upper() , ls))
```

```python
ls
```

```python
list(map(lambda x : x.upper() , ls))
```

```python
a = (map(lambda x : x.upper() , ls))
```

```python
for i in a:
    print(i)
```

```python
import sys
import time

def printLyrics():
    # Each tuple is (line, delay_per_char)
    lines = [
        ("I wanna da-", 0.06),
        ("I wanna ro-", 0.07),
        ("I wanna rock your body", 0.08),
        ("I wanna go", 0.08),
        ("I wanna go for a ride", 0.068),
        ("Hop in the music and", 0.07),
        ("Rock your body", 0.08),
        ("Rock that body", 0.069),
        ("come on, come on", 0.035),
        ("Rock that body", 0.05),
        ("((Rock your body))", 0.03),
        ("Rock that body", 0.049),
        ("come on, come on", 0.035),
        ("Rock that body", 0.08),
    ]

    # Delays after each line, same length as lines, to pause between lines
    delays_after_line = [0.5] * len(lines)

    for i, (line, char_delay) in enumerate(lines):
        for char in line:
            sys.stdout.write(char)
            sys.stdout.flush()
            time.sleep(char_delay)
        print()  # new line after each lyric line
        time.sleep(delays_after_line[i])

if __name__ == "__main__":
    printLyrics()
```
