# DBMS
    - DATABASE MANAGEMENT SYSTEM
    

    - Database
        - Database is used to store our data....

    - Database management system
        - DBMS is a software / Tool which help to store , modify , delete the data..
        - Data manipulation

    - Advantage of DBMS
        - Data Integrity
        - Data security
        - Data Backup
        - Data sharing
        - Data Accessbility
        - Data Redundancy

    - Types of DBMS
        - Hierachical
            - Parent > child
        - Network
            - Graphs
        - RDBMS
            - Relational Database Management System
            - Data stored in row and column format
            - Row x column = Table
            - column > attribute > field
            - Row > Record > Tuple
            - software
                - MYSQL , ORACLE
        - NOSQL
            - Data stored in unstructed format..
            - Data stored in json format
                - Key : value pair
            - software
                - mongodb , cassendra , reddis
        - OODBMS
            - Object oriented Database Management system
            - Data stored in 
                - class and object format
            - software
                - ORM
                    - Object Relational Mapping , db4


        - Types of languages in RDBMS
            - DDL
                - Data defintion language
                    - DDL queries are used to define the schema / structure
                    - Queries In DDL
                        - CREATE
                            - Used to create db or table
                        - ALTER
                            - Help to modify schema...
                        - DROP
                            - Delete the schema
                        - TRUNCATE
                            - Only delete the records not the schema

            - DML
                - Data Manipulation Language
                    - Helps to modify the data in the table...
                        - Queries in DML
                            - INSERT
                                - Help to insert the data in the table
                            - UPDATE
                                - Help to update the data in the table
                            - DELETE
                                - Delete the data from the table

            - DQL
                - Data Query Language
                    - Helps to retreive the data
                        - Query in DQL
                            - SELECT
                                - Helps to fetch out the data from the table
         
         - DCL
             - Data Control Language
                 - Helps to give and take back the permission on our db
                 - Queries in DCL
                     - Grant
                         - Give access / permission to the user...
                     - Revoke
                         - To take back the grant permission from the user


        - TCL
            - Transcation Control Language
                - Queries In TCL
                    - COMMIT
                        - Helps to save the changes...
                    - ROLLBACK
                        - Rollback the last changes...
                    - SAVEPOINT
                        - Helps to rollback to the previous tracscations...


        - Types of constraints:
            - Constraints
                - Restictions
                - Rules & Regulations

             - Check
                 - check validation
            - NOT NULL
                - Cant leave without inserting

            - Unique
                - No duplicate values allowed..

            - Primary key
                - used to define a column as unique
                - Not null and unique...


            - Foreign key
                - Helps to create a relation between multiple tables


            - Default
                - To set a default value..



            - ACID PROPERTY
                - Atomicity
                    - Trascation  > complete > Failed > but not in process
                - Consitency
                    - All rules should be valid...
                - Isolation
                    - Trnascation will be performed independently..
                - Durability
                    - Data always remain permanent...


            - CRUD Operations
                - CREATE
                - READ
                - UPDATE
                - DELETE
