```python
file = open("att.csv","w")
header = "date,topic"
s = int(input("Total student's in class"))
for i in range(1,s+1):
    value = input(f"Enter {i} name")
    header += f",{value}"
else:
    header += "\n"
    file.write(header)
    file.close()
```
