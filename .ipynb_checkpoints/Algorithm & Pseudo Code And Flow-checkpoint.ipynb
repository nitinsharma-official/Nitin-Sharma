## Algorithm a number is odd / even
       - Start the program
       - Take a input from user
       - If number is completely divisible by 2(num % 2 == 0) then print even
       - otherwise print odd
       - stop the program

    start
    a = 10
    a % 2 == 0 > even
    odd
    stop
