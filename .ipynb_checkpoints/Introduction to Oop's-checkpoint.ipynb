```python
class university:
    def __init__(self , name , sid , marks):
        self.name = name
        self.sid = sid       # attribute initialization / attribute binding
        self.marks = marks

    def show(self):
        print(f"Name > {self.name}")
        print(f"Sid > {self.sid}")
        print(f"Marks > {self.marks}")
```

```python
poornima = university("Ganeshu" , 1003, 94)
```

```python
poornima.show()
```

```python
s1 = university("Nitin" , 1001, 98)
s2 = university("Meenakshi", 1002, 97)
```

```python
s1.show()
s2.show()
```

```python
s1.__dict__.items()
```

```python
for key , value in s1.__dict__.items():
    print(f"{key} > {value}")
```

```python
for key , value in poornima.__dict__.items():
    print(f"{key} > {value}")
```

        - Create a class student
        - Takes 5 attribute Name , Age , M1 , M2 , M3
        - Create three method's:
            - show name
                - which will show name of student
            - show marks
                - which will show total marks
            - show percentage
                - which will show percentage

```python
class student:
    def __init__(self , name , age , m1 , m2 , m3):
        self.name = name
        self.age = age
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def show_name(self):
        print(f"Name > {self.name}")

    def show_age(self):
        print(f"Age > {self.age}")

    def show_marks(self):
        total_marks = self.m1 + self.m2 + self.m3
        print(f"Total Marks > {self.total_marks}")

    def show_percentage(self):
        total_marks = self.m1 + self.m2 + self.m3
        percentage = total_marks / 300 * 100
        print(f"Total Percentage > {percentage}")
        
```

```python
st1 = student("Nitin" , 21 , 98 , 97 , 96)
st2 = student("Abhishek" , 21 , 98 , 87 , 78)
```

```python
st1.show_name()
```

```python
st2.show_percentage()
```

```python
st1.show_percentage()
```

```python
for key , value in st1.__dict__.items():
    print(f"{key} : {value}")
```

```python
for i , j in st2.__dict__.items():
    print(f"{i}  :  {j}")
```

```python
class Temperature:
    def __init__(self, high, low):
        self.high = high
        self.low = low

    def show_temperature(self):
        for h, l in zip(self.high, self.low):
            print(f"High: {h}°C  |  Low: {l}°C")
```

```python
tem1 = Temperature("Summer" , "Winter")
tem2 = Temperature("Summer" , "Winter")
```

```python
tem1.show_temperature()
```

```python
class school:
    def __init__(self , name , *marks):
        self.name = name
        self.marks = marks

    
    def show_total_marks(self):
        total = sum(self.marks)
        print(f"Name: {self.name}")
        print(f"Total_Marks: {total}")
```

```python
s1 = school("Nitin" , 88,78,56,78,56,77)
```

```python
s1.show_total_marks()
```
