# Advance Database

    In MySQL, IF-ELSE and loops are primarily used within stored procedures, functions, or triggers.
    They are not directly available for use in standalone SQL queries or scripts.
    However, there are some alternatives and workarounds depending on what you're trying to achieve.

1. Using IF-ELSE Logic Without Procedures

syntax:

    IF(condition, value_if_true, value_if_false)

    example: 
    
    SELECT name, IF(age >= 18, 'Adult', 'Minor') AS status from table_name;

2. CASE Statement

    syntax:
                case
                    when condition1 then result1
                    when condition2 then result2
                    else default result
                end

    example:
            select name,
                   case
                       when age < 18 then 'minor'
                       when age >= 18 and age < 65 then 'adult'
                       else 'senior citizen'
                   end as col_name
            from table_name;
                       

### Procedures in MySQL

    A stored procedure in MySQL is a precompiled collection of SQL statements that are stored in the database.
    It can be invoked with a single call, allowing you to encapsulate complex logic and reuse it across multiple applications or queries.
    Procedures are written in SQL and can include control-flow statements like IF, LOOP, WHILE, and CASE.

#### Key Features of MySQL Procedures
Reusability: Procedures can be called multiple times, reducing code duplication.

Modularity: Encapsulates logic into a single unit, making code easier to maintain.

Performance: Precompiled and stored in the database, reducing parsing and compilation overhead.

Security: Permissions can be granted to execute procedures without exposing underlying tables.

Parameter Support: Procedures can accept input, output, and input-output parameters.

    proc with no parameter
    
    delimiter //
    create procedure proc_name()
    begin
        select * from table;
    end //

    call proc_name(); //

    proc with in parameter
    
    delimiter //
    create procedure proc_name (IN var_name dtype)
        begin
            query
        end //

    proc with out parameter
    
    DELIMITER //
    CREATE PROCEDURE GetEmployeeCount(OUT var_name dtype)
    BEGIN
        SELECT COUNT(*) INTO total FROM table_name;
    END //

### Triggers

    a trigger is a special kind of stored procedure that automatically runs (or "fires") when a specific event occurs in the database. These events are usually related to changes in the data, such as when you insert, update, or delete rows in a table.    
    
Think of a trigger as a "watchdog" that keeps an eye on your table and performs an action whenever something specific happens.

Triggers are tied to one of the following events:

- INSERT: Fires when a new row is added to a table.
- UPDATE: Fires when a row in a table is modified.
- DELETE: Fires when a row is removed from a table.

Triggers can run either BEFORE or AFTER the event

    syntax:
            CREATE TRIGGER trigger_name
            [BEFORE | AFTER] [INSERT | UPDATE | DELETE]
            ON table_name
            FOR EACH ROW
            BEGIN
            -- SQL statements to execute
            END;

    trigger_name: The name of the trigger.
    BEFORE or AFTER: Specifies when the trigger runs.
    INSERT, UPDATE, or DELETE: Specifies the event that fires the trigger.
    table_name: The table the trigger is associated with.
    FOR EACH ROW: Ensures the trigger runs for each affected row.
    BEGIN ... END: Contains the SQL statements to execute.
