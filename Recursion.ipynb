## Recursion
        - When a function call itself again and again for solving a specific task..

#### Two type of approach for solving a problem
        - Iterative approach
               - using loops
        - Recursive approach
               - using fuctions

```python
def fun():
    print("Hello user")
    fun()
fun()
```

```python
a = 1
def fun1():
    global a
    print("Hello World")
    a += 1
    fun()
```

```python
def fun(n):
    if n == 0:
        return
    print("Hello user")
    return fun(n-1)
```

```python
fun(5)
```

```python
def add(n):
    if n == 5:
        return 5
    return n + add(n+1)
```

```python
add(1)
```

```python
def fact(c):
    if c == 0 or c == 1:
        return 1
    return c * (fact(c - 1))
```

```python
fact(5)
```

```python
def fact1(e):
    if e == 0 or e == 1:
        return 1
    return e * (fact1(e - 1))
```

```python
fact1(10)
```
